```python
import pandas as pd
from sklearn.cluster import KMeans,DBSCAN
import numpy as np
import matplotlib.pyplot as plt

# Sample data

data_points = np.array([[1, 2], [2, 3], [2, 4], [3, 1], [3, 3], [4, 4], [4, 7], [5, 9], [6, 4], [6, 5], [6, 8], [7, 6], [8, 8], [8, 9], [9, 4]])
df = pd.DataFrame(data_points, columns=['x', 'y'])
dbscan = DBSCAN(eps=2, min_samples=3)
dbscan.fit(data_points)

# Add the cluster labels to the dataframe
df['cluster'] = dbscan.labels_

# Plot the clustered data
plt.scatter(df['x'], df['y'], c=df['cluster'], cmap='viridis')
plt.xlabel('x')
plt.ylabel('y')
plt.title('DBSCAN Clustering')
plt.show()
```

```python
def find_centroid(cluster_points):
    return np.mean(cluster_points, axis=0)

# Example usage:
cluster_points = [[4,7],[5,9],[6,8],[7,6],[8,8],[8,9]]
centroid = find_centroid(cluster_points)
print(centroid)
```
