```python
import numpy as np 
from sklearn import svm
from mlxtend.plotting import plot_decision_regions
```

```python
training_data = { 
    "x1" : [3.5,5,3,5.5,4.5,3.5,3,4],
    "x2" :[-2,1.5,3,-1,-5,-6.5,-4,-3.5],
    "y" : [1,1,1,1,-1,-1,-1,-1]
}
x = list(zip(training_data["x1"], training_data["x2"]))
y = training_data["y"]
clf = svm.SVC(kernel='linear')
d =clf.fit(x,y)
alphas = np.abs(d.dual_coef_)
support_vectors = d.support_vectors_
bias = d.intercept_
print("alphas",alphas)
print("support vectors",support_vectors)
print("parameter weight w :",d.coef_)
print("parameter bias b",bias)
```

```python
testing_data = { 
    "x1" : [5.5,3.5,4.5,2.5,3.5,2.5,4,5],
    "x2" :[0.5,-0.5,1.5,-1,-4.,-4,-6,-5],
    "y" : [1,1,1,1,-1,-1,-1,-1]
}
x = list(zip(training_data["x1"], training_data["x2"]))
# y = training_data["y"]
for test_data in x:
    print("prediction for ",test_data," is ",d.coef_@test_data+bias)
```
